```python
%sql
USE CATALOG databricks_medallion;
USE SCHEMA shop;
```

```python
sales=spark.table("sales_silver")
region=spark.table("regions_silver")
```

```python
sales_with_region=sales.join(region,on="region",how="inner")
sales_with_region.display()
```

```python
currency=spark.table("currency_silver")
sales_full=sales_with_region.join(currency,on='currency',how='inner')
sales_full.display()
```

```python
from pyspark.sql.functions import count,sum
df_sales=spark.table("sales_silver")
df_sales_summary=df_sales.groupBy("region").agg(count("*").alias("total_orders"),sum("amount_usd").alias("total_sales_usd"))
```

```python
df_sales_summary.write.format("delta").mode("overwrite").saveAsTable("sales_gold")
```

```python
df_products=spark.table("sales_silver")
df_products_summary=df_products.groupBy("product").agg(count("*").alias("units_sold"),sum("amount_usd").alias("total_sales"))
```

```python
df_products_summary.write.format("delta").mode("overwrite").saveAsTable("products_gold")
```
