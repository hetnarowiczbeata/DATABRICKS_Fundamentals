```python
from pyspark.sql.functions import concat,lit,col
```

```python
from pyspark.sql.functions import concat, lit, col
df_regions=spark.table("sales_filtered").select("region").distinct().withColumn("description",concat(lit("Region "),col("region")))
df_regions.write.format("delta").mode("overwrite").saveAsTable("regions_silver")
```

```python
spark.table("regions_silver").display()
```

```python
from pyspark.sql.functions import max

df_products=spark.table("sales_filtered").groupBy("product").agg(max("amount_usd").alias("max_price")).withColumn("description",concat(lit("Product "),col("product")))
```

```python
df_products.write.format("delta").mode("overwrite").saveAsTable("products_silver")
spark.table("products_silver").display()
```

```python
df_currency=spark.table("sales_filtered").select("currency").distinct().withColumn("description",concat(lit("Currency "),col("currency")))
df_currency.write.format("delta").mode("overwrite").saveAsTable("currency_silver")
spark.table("currency_silver").display()                                                                                  
```
