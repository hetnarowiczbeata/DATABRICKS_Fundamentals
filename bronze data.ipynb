```sql
%python
# Import spark session
from pyspark.sql import SparkSession
#Create spark session
spark=SparkSession.builder.getOrCreate()
```

```sql
%python
spark.sql("USE CATALOG databricks_medallion")
spark.sql("USE SCHEMA shop")
```

```sql
%python
df= spark.read.format("csv")\
    .option("header","true")\
    .option("inferSchema","true")\
    .load("/Volumes/databricks_medallion/shop/flat_files/orders.csv")
```

```sql
%python
df.display()
```

```sql
%python
from pyspark.sql.functions import col

df_clean=df.withColumnRenamed("amount","amount_usd").withColumn("amount_usd",col("amount_usd").cast("double"))
```

```sql
%python

df_clean.write.format("delta").mode("overwrite").saveAsTable("sales_bronze")
```
