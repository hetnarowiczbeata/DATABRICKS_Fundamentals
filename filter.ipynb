```python
from pyspark.sql import SparkSession

spark=SparkSession.builder.getOrCreate()
```

```python
df=spark.table("sales_bronze")
df.select("product","amount_usd").display()
```

```python
dfEurope=df.filter(df.region=="Europe").display()
```

```python
dfAsiaabove500=df.filter((df.region=="Asia") &(df.amount_usd>500)).display()
```

```python
from pyspark.sql.functions import when

dfHV=df.withColumn("high value",when(df.amount_usd>500,"Yes").otherwise("No")).display()
```

```python
dfAE=df.filter(df.region.isin("Asia","Europe")).display()
dfsmart=df.filter(df.product.startswith("Smart")).display()
```

```python
dfHV=df.write.format("delta").saveAsTable("sales_filtered")
```
